```python
l1=[23,56,43,43,23,45,67]

def sor(l):
  n = len(l)
  for i in range(n-1):
    for j in range(n-i-1):
      if l[j] > l[j+1]:
        l[j], l[j+1] = l[j+1], l[j] # Swap elements
  return l

print(sor(l1))
```
